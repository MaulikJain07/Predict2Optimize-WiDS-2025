# **Problem 0**


#### (a)
$$
\begin{aligned}
\min_{x \in \mathbb{R}^n} \quad & x^\top Q x + c^\top x \\
\text{s.t.} \quad & Ax = b
\end{aligned}
$$

where $Q \succeq 0$, $A \in \mathbb{R}^{m \times n}$, $b \in \mathbb{R}^m$.



--> The given problem is convex because the  $  x^\top Q x $ is convex only when Q is PSD, which it is given as Q>=0 means all the eigen values are >=0. The linear part is always convex..

#### (b)
$$
\begin{aligned}
\min_{x \in \mathbb{R}^n} \quad & \|x\|_2^2 \\
\text{s.t.} \quad & \|Ax\|_2 \ge 1
\end{aligned}
$$
where $A \in \mathbb{R}^{m \times n}$.

--> The given problem is not convex, because of the constraint part. However, the objective is a convex function. But the constraint is >= for a convex function, it should be <= or the function should have been concave.

#### (c)
$$
\begin{aligned}
\max_{w \in \mathbb{R}^n} \quad & (\mu^\top w)(\mathbf{1}^\top w) \\
\text{s.t.} \quad & w \ge 0
\end{aligned}
$$

where $\mu \in \mathbb{R}^n$ is fixed.

--> It is not, because the given problem is non-convex maximisation problem.
